```python
import numpy as np
import networkx as nx
import matplotlib.pylot as plt
from collections import deque
```

```python
connections = np.array([(1,2),
                       (1,3),
                       (1,8),
                       (1,9)
                      ],dtype = int)
direct ={}
# Convert matrix to list of edge of each node
for i in range(len(connections)):
    if connections[i][0] in direct:
        direct[connections[i][0]].append(connections[i][1])
    else:
        direct[connections[i][0]] = [connections[i][1]]

print(direct)
```

```python
# List the set of direction of one node
def directory(connections):
    direct ={}
    for i in range(len(connections)):
        if connections[i][0] in direct:
            direct[connections[i][0]].append(connections[i][1])
        else:
            direct[connections[i][0]] = [connections[i][1]]

    return direct
```

```python

def illustration(connections):
    G = nx.DiGraph()
    
    # formalize
    direction = directory(connections)
    
    # Add edges to graph
    for i in direction:
                            
        for j in direction[i]:
            G.add_edge(i,j)
    # choose kind of layout for node
    pos = nx.circular_layout(G)
    
    # draw nodes and edges
    nx.draw_networkx_nodes(G,pos,node_size = 600)
    nx.draw_networkx_edges(G,pos,connections,width = 3,style = "dashed")
    # Node labels
    nx.draw_networkx_labels(G,pos,font_size=15,font_family="sans-serif")
    
    ax = plt.gca()
    ax.margins(0.008)#prevent the plot touching the frame
    plt.axis("off")#hide axes
    plt.tight_layout() # draw without coordiante graph
    plt.show()
# 5 nodes
nNode = 8
connections = np.array([
                        [1,2],
                        [1,3],
                        [1,4],
                        [6,7],
                        [6,5]
                       ])

# illustration(connections)
```

```python
nNode = 7
connections = np.array([
                        [1,2],
                        [1,3],
                        [1,4],
                        [6,7],
                        [6,5]
                       ])

class connection2matrix:
    def __init__(self,connections = None,num = 0):
        self._nNode = num
        self._matrix = np.zeros((self._nNode,self._nNode))
        
        for i in connections:
            self._matrix[i[0]-1][i[1]-1] = 1
    def illustration(self):
        G = nx.DiGraph()
        # convert data to matrix
        for i in range(self._nNode):
            for j in range(self._nNode):
                if abs(self._matrix[i][j]-1) < 1e-7:
                    G.add_edge(i+1,j+1)
                    
        pos = nx.circular_layout(G)

        nx.draw_networkx_nodes(G,pos,node_size=600)
        nx.draw_networkx_edges(G,pos,connections,width = 3,style = 'dashed')

        nx.draw_networkx_labels(G,pos,font_size=15,font_family="sans-serif")

        ax = plt.gca()
        ax.margins(0.008)#prevent the plot touching the frame
        plt.axis("off")#hide axes
        plt.tight_layout() # draw without coordiante graph
        plt.show()

result = connection2matrix(connections,nNode)
result.illustration()
```
