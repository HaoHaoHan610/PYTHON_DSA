```python
def ruler(n: int):
    if n == 0:
        return "--------0"
    return ruler(n-1) + f"\n----\n----\n----\n----\n--------{n}"
                        

print(ruler(6))
```

```python
def draw_DashedLine(num: int, label: str = ''):
    line = '-' * num
    if label:
        line += ' ' + label
    print(line)

def draw_interval(num:int):
    if num > 0:
        draw_interval(num-1)
        draw_DashedLine(num)
        draw_interval(num - 1)
def draw_Ruler(num: int,length:int):
    draw_DashedLine(num,str(0))

    for i in range(1,num+1):
        draw_interval(length - 1)
        draw_DashedLine(num,str(i))

draw_Ruler(7,4)
```

```python
def binarySearch(left,right,array,target):
    mid = (left + right)//2
    print(mid)
    if left < right:
        if array[mid - 1] == target:
            return mid - 1
        if array[mid - 1] > target:
            binarySearch(left,mid - 1,array,target)
        else:
            binarySearch(mid + 1, right,array,target)
    return None

data = [1,2,4,9,10,11]
print(binarySearch(1,len(data),data,3))        
```

```python
def bad_power(x,n):
    if n == 0:
        return 1
    return x*power(x,n-1)


def power(x,n):
    if  n==0: return 1
    ans = power(x,n//2) # -> 
    result = ans*ans
    if n%2 == 1:
        result *= x
    return result
```

```python
from time import time
print("BAD POWER")
start = time()
print(str(bad_power(2,10000)) + " Take time: "+str(time()-start))
print("Good one")
start = time()
print(str(power(2,10000)) +" Take time: "+str(time()-start))
```

# BINARY SUM


```python
def binary_sum(data,start,stop):
    if start > stop:
        return 0
    elif start == stop - 1:
        return data[start]
    else:
        mid = (stop - start )//2
        return binary_sum(data,start,mid) + binary_sum(data,mid, stop)
```

```python
import copy
def PuzzleSolve(s,v,u):
    for k in u:
        # print(k)
        v.append(k)
        u.remove(k)
        PuzzleSolve(s-1,v,u)
        if s == 1:
            print("Solution: ",end='')
            print(v)
        u.append(k)
        v.remove(k)
```

```python
data = [1,2,3,4]
PuzzleSolve(len(data),[],data)
```
