````python
def findMax(position,maxValue,data):
    if position == len(data):
        return maxValue
    if data[position] > maxValue:
        maxValue = data[position]
    return findMax(position + 1,maxValue,data)

data = [1,2,3,34]


print(findMax(position=0,maxValue=data[0],data = data))
````

````python
def Power(x,num):
    if num == 0:
        return 1
    temp = Power(x,num//2)
    ans = temp*temp
    if num%2 == 1:
        ans *= x
    return ans



print(Power(2,10))
````

````python
# import swap

def reverse(low,high,data):
    if low == high - 1:
        return
    data[low],data[high-1] = data[high-1],data[low]
    reverse(low + 1, high -1,data)

data = [4,3,6,2,6]

reverse(0,len(data),data)
print(data)
````

````python
def isPalindrome(s,low,high):
    if low == high:
        return True
    if s[low]!=s[high]:
        return False
    else:
        return True
    return isPalindrome(s,low + 1,high - 1)

s = 'connoc'
isPalindrome(s,0,len(s) - 1)    
````

````python
vowels = ['a','o','e','u','i']

def isMoreVowels(s,low,high,count):
    if  low >= high or high < 0 or low > len(s)-1:
        if low == high and s[low] in vowels:
            count+=1
        return count > len(s)//2
    if s[low] in vowels:
        count +=1
    if s[high] in vowels:
        count += 1
    return isMoreVowels(s,low+1,high-1,count)
s = 'aoeunh'
print(isMoreVowels(s,0,len(s) - 1,0))
````

````python
vowels = {'a','o','e','u','i'}
# Use the time to look up is log(n)
def isMoreVowelsPrime(s,low,high):
    if low > high:
        return 0
    if low == high:
        return 1 if s[low] == s[high] else 0
    current_count = 0
    if s[high] in vowels:
        current_count += 1
    if s[low] in vowels:
        current_count += 1

    return current_count + isMoreVowels(s,low+1,high-1)
````

````python
def drawline(length: int, label: str = "") -> None:
    line = "-" * length
    if label:
        line += " " + label
    print(line)

def draw_interval_iter(n: int) -> None:
    for i in range(1, 1 << n):
        length = (i & -i).bit_length()
        drawline(length)

def draw_ruler(num_inches: int, major_length: int) -> None:
    drawline(major_length, "0")
    for inch in range(1, num_inches + 1):
        draw_interval_iter(major_length - 1)
        drawline(major_length, str(inch))

draw_ruler(7, 4)
````
