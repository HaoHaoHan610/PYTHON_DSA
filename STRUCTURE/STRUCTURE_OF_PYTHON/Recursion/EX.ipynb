```python
def findMax(position,maxValue,data):
    if position == len(data):
        return maxValue
    if data[position] > maxValue:
        maxValue = data[position]
    return findMax(position + 1,maxValue,data)

data = [1,2,3,34]


print(findMax(position=0,maxValue=data[0],data = data))
```

```python
def Power(x,num):
    if num == 0:
        return 1
    temp = Power(x,num//2)
    ans = temp*temp
    if num%2 == 1:
        ans *= x
    return ans



print(Power(2,10))
```

```python
# import swap

def reverse(low,high,data):
    if low == high - 1:
        return
    data[low],data[high-1] = data[high-1],data[low]
    reverse(low + 1, high -1,data)

data = [4,3,6,2,6]

reverse(0,len(data),data)
print(data)
```

```python
def PuzzleSolve(num,S,U):
    for i in U:
        S.append(i)
        # print("S now")
        # print(S)
        U.remove(i)
        # print("U now")
        # print(U)
        
        if num == 1:
            print(S)

        else:
            PuzzleSolve(num - 1,S,U)
        S.remove(i)
        U.append(i)

data = ['a','b','c','d']

PuzzleSolve(3,[],data)
```

```python
def HarnomicNum(i,num):
    convert = int(num[i])-int('0')
    
    if(i == 1):
        return convert
    
    # print(convert*pow(10,len(num)-1))
    return convert*pow(10,len(num)-1) + HarnomicNum(i + 1,num)

print(HarnomicNum(0,'10000'))
    
```
