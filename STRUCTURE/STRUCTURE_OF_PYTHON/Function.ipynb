# *FUNCTION*

- **Function:**  a traditional stateless function that is invoked without invoked.
- **Method:** a member function that is invoked upon a specific object using object-oriented message passing syntax

- **Activation record:** Store information relevant to the current call
    - **Namespace:** to manage all identifier
    - **Local scope:** to within the current call
