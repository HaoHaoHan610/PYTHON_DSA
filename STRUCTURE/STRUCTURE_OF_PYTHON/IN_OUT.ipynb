* Generator: the technique for creating the *interator*
    * *yield:* Indecate each element of the series -> use when have work with large data, pipeline

```python
num = 100
def fac(n):
    for i in range(1,n + 1):
        if n % i == 0:
            yield i
l = list(fac(100))
print(l)
```

```python
def factors(n): # generator that computes factors
    k = 1
    while k * k < n: # while k < sqrt(n)
        if n % k == 0:
            yield k
            yield n // k
        k += 1
    if k * k == n: # special case if n is perfect square
        yield k
l = list(fac(100))
print(l)
```
