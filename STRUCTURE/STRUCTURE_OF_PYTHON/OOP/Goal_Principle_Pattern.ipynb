# **VECTOR**

```python
class vector:
    """Vector"""
    def __init__(self,d):
        "d: dimension of vector"
        self._coordinarates = [0]*d

    def __len__(self):
        return len(self._coordinarates)

    def __getitem__(self,k):
        if k > len(self._coordinarates) - 1 or k < 0:
            raise IndexError(f'Dimension is {len(self._coordinarates)}')
        return self._coordinarates[k]

    def __setitem__(self,k,val):
        if k > len(self._coordinarates) - 1 or k < 0:
            raise IndexError(f'Dimension is {len(self._coordinarates)}')
        self._coordinarates[k] = val

    def __add__(self,other):
        if len(self) != len(other):
            raise ValueError('Dimensions dont fit each other')
        result = vector(len(self))
        for j in range(len(self)):
            result[j] = self[j] + other[j]
        return result

    def __eq__(self,other):
        return self._coordinarates == other._coordinarates

    def __ne__(self,other):
        return not self == other

    def __str__(self):
        return '<' + str(self._coordinarates)[1:-1] + '>'
    
```

# **Iterator**

```python
class SequenceIterator:
    def __init__(self,sequence):
        self._seq = sequence
        self._k = -1

    def __next__(self):
        self._k += 1
        if self._k < len(self._seq):
            return (self._seq[self._k])
        else:
            raise StopIteration()

    def __iter__(self):
        return self
```

```python
class Range:
    def __init__(self,start,stop = None, step = 1):
        if step == 0:
            raise ValueError('Step cannot be 0')
        if step is None:
            start, stop = 0, start
        self._length = max(0,(stop - start + step -1)//step)
        self._start = start
        self._step = step


    def __len__(self):
        return self._length

        
    def __getitem__(self,k):
        if k < 0:
            k+= len(self)
        if not 0 <= k < self._length:
            raise IndexError("Index out of range")

        return self._start + k*self._step
list(Range(0,100))  
        
```
