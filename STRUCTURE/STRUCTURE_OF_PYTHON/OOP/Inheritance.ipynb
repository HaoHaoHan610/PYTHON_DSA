# Abstract base classes

- **Abstract base classes:** can guarantee one or more **abstract methods** when conrete class is hard to define for totally super classes
- Support for polymorphism

- *Metaclass:* the class defination itself
- *@abstracmethod:* declare -> it forces super class must definate its method again
### Metaclass -> class -> object
## a better module -> collerctions

```python
from abc import ABCMeta,abstractmethod

class sequence(metaclass = ABCMeta):

    @abstractmethod
    def __len__(self):
        """Return length of the sequence."""

    @abstractmethod
    def __getitem__(self,j):# v[j]
        """Return the element at index j for the sequence."""
        # it forces to override in the inherated superclass

    def __contain__(self,val): # v in a
        for j in range(len(self)):
            if self[j] == val:
                return True
        return False

    def index(self,val):
        for j in range(len(self)):
            if self[j] == val:
                return j
        raise ValueError('Value not in sequence')

    def count(self,val):
        k = 0
        for j in range(len(self)):
            if self[j] == val:
                k+=1
        return k
```

# *'@':* method uses to wrap the function or class (protection)

```python
def debug(f):
    def wrapper(*args, **kwargs):
        try:
            print("Gọi hàm:", f.__name__)
            result = f(*args, **kwargs)
            print("Kết quả:", result)
            return result
        except:
            raise ValueError('Wrong type of data')
    return wrapper

@debug
def sum(a,b):
    return a + b

sum('ngu','huh')
```

# Shallow and Deep Copying

- if we change the instance namspace change totally original stored value
- That is called **shallow copy**

```python
palette = ['orange','purple','red','blue']
warmtones = palette
print(palette)
print(warmtones)
palette[0] = 'white'
print(palette)
print(warmtones)
```

```python
import copy #module of deep copy
palette = ['orange','purple','red','blue']

warmtones = copy.deepcopy(palette) #use deepcopy

print(palette)
print(warmtones)
palette[0] = 'white'
print(palette)
print(warmtones)
```
