# C-1.15


```python
def checkDistinctNumber(data):
    for i in range(0,len(data)):
        for j in range(i + 1,len(data)):
            if data[i] == data[j]:
                return False
    return True

def set_checkDistinctNumber(data):
    return len(set(data)) == len(data)

data = [1,2,4,5,5,6,7]

set_checkDistinctNumber(data)
```

```python
import random as rd
def self_Shuffle(data):
    l = []
    while(len(data)):
        index = rd.randint(0,len(data)-1)
        l.append(data[index])
        del data[index]
    return l


data = [1,4,5,6,8]
l = self_Shuffle(data)
print(l)
    
```

```python
def product(a,b):
    l = []
    for i in range(len(a)):
        l.append(a[i]*b[i])
    return l

a = [1,2,3,4]
b = [4,3,2,1]

product(a,b)
```

```python
def product(a,b):
    return [x*y for x,y in zip(a,b)]
# Zip makes arrays into a tuple
a = [1,2,3,4]
b = [4,3,2,1]

product(a,b)
```

```python
from random import randint
import random
import string
def typos(s):
    vowels = [i for i in string.ascii_lowercase]
    # convert alphabet to list of letters 
    index = random.randint(0,len(s)-1)
    while s[index] == ' ' :
        index = random.randint(0,len(s)-1)
    s = s.replace(s[index], random.choice(vowels))
    #replace randomly
    return s 

def writeOut(s,n,typosNumber):
    l = [s for i in range(n)]
    for i in range(typosNumber):
        l[i] = typos(l[i])
    random.shuffle(l) # shuffle all stuffs
    with open("punishment.txt",'a') as f:
        for i in l:
            f.write(i + '\n')
            # write down file
writeOut("I will never spam my friends again.",100,8)     
# s = "hao hao han"
# s[0]
```
