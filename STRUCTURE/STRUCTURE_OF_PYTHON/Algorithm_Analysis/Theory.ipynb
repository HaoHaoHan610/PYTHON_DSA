# ***BIG-OH***

# $ f(n) \text{ is a } O(g(n))$
> ## IF
> $f(n) \leq c.g(n)$ when $n \geq n_0$

> ## EXAMPLE
> 

# $ f(n) \text{ is a } \Omega (g(n))$
> ## if
> $f(n) \geq c.g(n)$ when $n \geq n_0$

> # *Example*
> ### $$3n \log n - 2n \leftrightarrow n \log n + 2(\log n - 1) \leq n \log n $$
> ### if only if
> $$ n_0 \geq 2 \text{ and } c = 1$$

# Big-Theta
> ## IF $$ \Theta(g(n)) $$
> $$c'g(n) \leq f(n) \leq c''g(n)$$
